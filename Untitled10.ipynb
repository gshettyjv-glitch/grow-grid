```python

import pandas as pd

# Load your CSV
df = pd.read_csv("C:\\Users\\gshet\\OneDrive\\Documents\\amazon_prime_video_data.csv)

# Fill missing values:
# - Numeric columns → mean
# - Categorical columns → mode
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:   # numeric columns
        df[col] = df[col].fillna(df[col].mean())
    else:                                       # categorical/object columns
        df[col] = df[col].fillna(df[col].mode()[0])

# Verify no missing values remain
print(df.isnull().sum())
```

```python
# Select a single column
print(df['ColumnName'])

# Select multiple columns
print(df[['Column1', 'Column2']])

# Filter rows (example: values greater than 100)
print(df[df['ColumnName'] > 100])

# Group by and aggregate
print(df.groupby('CategoryColumn')['NumericColumn'].mean())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\gshet\OneDrive\Desktop\sales_customer_analytics.csv")
data = df.groupby('Product')['OrderID'].count()

plt.bar(data.index, data.values)
plt.title("Orders by Product")
plt.xlabel("Product")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()

```

```python
data = df.groupby('Product')['UnitPrice'].sum()

plt.bar(data.index, data.values)
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
data = df.groupby('Product')['UnitPrice'].sum()

plt.bar(data.index, data.values)
plt.title("Revenue Concentration")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
data = df.groupby('Product')['UnitPrice'].sum().sort_values(ascending=False)
data = data.cumsum()

plt.plot(data.index, data.values, marker='o')
plt.title("Cumulative Revenue")
plt.xlabel("Product")
plt.ylabel("Cumulative Revenue")
plt.xticks(rotation=45)
```

```python
data = df.groupby('Product').agg(
    Revenue=('UnitPrice', 'sum'),
    Demand=('Quantity', 'sum')
)

plt.scatter(data['Demand'], data['Revenue'])
plt.title("Revenue vs Demand")
plt.xlabel("Demand")
plt.ylabel("Revenue")
plt.show()
```

```python

data = df.groupby('Product')['UnitPrice'].std()

plt.bar(data.index, data.values)
plt.title("Price Variation by Product")
plt.xlabel("Product")
plt.ylabel("Price Variation")
plt.xticks(rotation=45)
plt.show()
```

```python
df.boxplot(column='UnitPrice', by='Product')

plt.title("Price Distribution")
plt.xlabel("Product")
plt.ylabel("Unit Price")
plt.xticks(rotation=45)
plt.show()

```

```python
df['PriceBand'] = pd.qcut(
    df['UnitPrice'], 3,
    labels=['Low', 'Medium', 'High']
)

data = df.groupby('PriceBand', observed=False)['NetSales'].sum()

plt.bar(data.index, data.values)
plt.title("Net Sales by Price Band")
plt.xlabel("Price Band")
plt.ylabel("Net Sales")
plt.show()
```

```python
data = df.groupby('PriceBand', observed=False)['OrderStatus'].apply(
    lambda x: x.isin(['Cancelled', 'Returned']).mean() * 100
)

plt.bar(data.index, data.values)
plt.title("Cancellation and Return Rate")
plt.xlabel("Price Band")
plt.ylabel("Rate (%)")
plt.show()
```

```python

data = df.groupby('PriceBand', observed=False)['OrderStatus'].apply(
    lambda x: x.isin(['Cancelled', 'Returned']).mean() * 100
)

plt.bar(data.index, data.values)
plt.title("Problem Orders by Price Band")
plt.xlabel("Price Band")
plt.ylabel("Rate (%)")
plt.show()
```

```python
data = df.groupby('Quantity')['OrderStatus'].apply(
    lambda x: (x == 'Delivered').mean() * 100
)

plt.bar(data.index, data.values)
plt.title("Delivery Success by Quantity")
plt.xlabel("Quantity")
plt.ylabel("Success Rate (%)")
plt.show()
```

```python
data = df.groupby('Quantity')['NetSales'].mean()

plt.bar(data.index, data.values)
plt.title("Average Net Sales by Quantity")
plt.xlabel("Quantity")
plt.ylabel("Average Net Sales")
plt.show()
```

```python
data = df.groupby('Quantity')['DiscountPct'].mean()

plt.bar(data.index, data.values)
plt.title("Discount by Quantity")
plt.xlabel("Quantity")
plt.ylabel("Average Discount")
plt.show()
```

```python
data = df.groupby('Product')['Quantity'].mean()

plt.bar(data.index, data.values)
plt.title("Quantity Purchased by Product")
plt.xlabel("Product")
plt.ylabel("Average Quantity")
plt.xticks(rotation=45)
plt.show()
```

```python
data = df.groupby('Product').agg(
    Discount=('DiscountPct', 'mean'),
    Demand=('Quantity', 'sum')
)

plt.scatter(data['Discount'], data['Demand'])
plt.title("Discount vs Demand")
plt.xlabel("Average Discount")
plt.ylabel("Demand")
plt.show()
```

```python
data = df[df['PriceBand'] == 'High'].groupby('CustomerCity')['Quantity'].sum()

plt.bar(data.index, data.values)
plt.title("Premium Product Demand by City")
plt.xlabel("City")
plt.ylabel("Total Quantity")
plt.xticks(rotation=45)
plt.show()
```

```python
data = df.groupby('CustomerCity')['DiscountPct'].mean()

plt.bar(data.index, data.values)
plt.title("Average Discount by City")
plt.xlabel("City")
plt.ylabel("Average Discount")
plt.xticks(rotation=45)
plt.show()
```

```python
ata = pd.crosstab(df['CustomerCity'], df['PaymentMode'])

data.plot(kind='bar')
plt.title("Payment Preference Across Cities")
plt.xlabel("City")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
```
